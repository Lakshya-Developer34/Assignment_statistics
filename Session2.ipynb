```python
# Task 3
youtube = [100, 120, 110, 105, 5000]
youtube2 = [100,110,110,130,150]
print("Mean 1 is " , st.mean(youtube))
print("Median 1 is" , st.median(youtube))
print("Mode 1 is " , st.mode(youtube))

print("-------------------------------------")

print("Mean 2 is " , st.mean(youtube2))
print("Median 2 is" , st.median(youtube2))
print("Mode 2 is " , st.mode(youtube2))
```

```python
import pandas as pd

# Load the Kaggle dataset
df = pd.read_csv("IPL2023_Match_Scoreboard.csv")

# Select the numeric column
runs = df["Runs"]

# Calculate mean, median and mode
mean_runs = runs.mean()
median_runs = runs.median()
mode_runs = runs.mode()

print("Mean:", mean_runs)
print("Median:", median_runs)
print("Mode:", mode_runs)
```
