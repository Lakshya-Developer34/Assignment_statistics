```python
# Task 1
def bayes_posterior(prior, likelihood, evidence):
    posterior = (likelihood * prior) / evidence
    return posterior


# Example
prior = 0.10
likelihood = 0.80   
evidence = 0.26

result = bayes_posterior(prior, likelihood, evidence)

print("Posterior probability:", result)
print("Percentage:", result * 100, "%")
```

```python
# Task 2
def bayes_posterior(prior, likelihood, evidence):
    return (likelihood * prior) / evidence


# Prior: 10% restaurants are premium
prior_premium = 0.10

# Likelihood: 80% of premium restaurants offer discounts
likelihood_discount_given_premium = 0.80

# Probability of discount:
# (Premium × Discount) + (Regular × Discount)
evidence_discount = (
    0.10 * 0.80 +
    0.90 * 0.20
)

posterior = bayes_posterior(
    prior_premium,
    likelihood_discount_given_premium,
    evidence_discount
)

print("Probability restaurant is premium:", posterior)
print("Percentage:", posterior * 100, "%")
```

```python
def bayes_posterior(prior, likelihood, evidence):
    return (likelihood * prior) / evidence


# Prior probability
p_spam = 0.50
p_not_spam = 0.50

# Probability that "free" appears
# in spam and non-spam emails
p_free_given_spam = 0.80
p_free_given_not_spam = 0.10

# Evidence: overall probability of "free"
p_free = (
    p_free_given_spam * p_spam +
    p_free_given_not_spam * p_not_spam
)

# Bayes theorem
p_spam_given_free = bayes_posterior(
    p_spam,
    p_free_given_spam,
    p_free
)

print("P(Spam | Free):", p_spam_given_free)
print("Percentage:", p_spam_given_free * 100, "%")
```

```python
# Task 5
def bayes_posterior(prior, likelihood, evidence):
    return (prior * likelihood) / evidence


# Prior probability:
# 40% of users follow this type of account
p_follow = 0.40

# 70% of followers like the post
p_like_given_follow = 0.70

# 60% do not follow the account
p_not_follow = 0.60

# 20% of non-followers like the post
p_like_given_not_follow = 0.20

# Overall probability of liking the post
p_like = (
    p_like_given_follow * p_follow +
    p_like_given_not_follow * p_not_follow
)

# Calculate P(Follow | Like)
result = bayes_posterior(
    p_follow,
    p_like_given_follow,
    p_like
)

print("P(Follows | Likes):", result)
print("Percentage:", result * 100, "%")
```
