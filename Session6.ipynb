1. Null and Alternative Hypothesis

Scenario: You believe people spend more time on Instagram than YouTube daily.

1 H₀ (Null Hypothesis): People spend the same amount of time on Instagram and YouTube daily.
  H₀: μInstagram ≤ μYouTube
2 H₁ (Alternative Hypothesis): People spend more time on Instagram than YouTube daily.
  H₁: μInstagram > μYouTube

Since the claim is "more than", this is a one-tailed test.

```python
# pip install scipy
```

```python
from scipy.stats import ttest_ind

mean_A = 350
mean_B = 320
sd_A = 50
sd_B = 50
n_A = 30
n_B = 30

# Create equivalent data approximately for demonstration
import numpy as np

np.random.seed(42)

group_A = np.random.normal(mean_A, sd_A, n_A)
group_B = np.random.normal(mean_B, sd_B, n_B)

t_stat, p_value = ttest_ind(group_A, group_B)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

3. A/B Test Example — WhatsApp Dark Mode

Let's test a new dark-mode design.

Control Group (A)

Users see the old/current design.

Variant Group (B)

Users see the new dark-mode design.

Randomly divide users into two groups, for example:

Group A → 5,000 users
Group B → 5,000 users
Metric

We could measure daily app engagement, such as average time spent in the app or the percentage of users who return the next day.

Hypotheses

H₀: The new dark mode does not change user engagement.

H₁: The new dark mode increases user engagement.

After collecting enough data, calculate the p-value.

p-value < 0.05 → Reject H₀ → evidence that the new design has an effect.
p-value ≥ 0.05 → Fail to reject H₀ → not enough evidence of an effect.


4. Flipkart Wishlist Button A/B Test

Given:

New button: 520 clicks
Old button: 480 clicks
Total = 1,000 users

The difference is:

520 − 480 = 40 users, or 4 percentage points.

At first glance, the new button looks better, but a difference doesn't automatically mean it is statistically significant.

The p-value tells us how likely it would be to see a difference this large (or larger) just because of random chance if there were actually no difference.

Simple decision rule
p-value < 0.05 → statistically significant → Reject H₀
p-value ≥ 0.05 → not statistically significant → Fail to reject H₀

For this particular 520 vs. 480 split, the difference is not large enough to be statistically significant at the 0.05 level in a simple two-proportion comparison.

Conclusion: The new button has a slightly higher click rate, but we don't have enough evidence to say the improvement isn't simply due to chance.
