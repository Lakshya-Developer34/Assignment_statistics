# Task 4 Slope and Intercept — Flipkart Price Prediction

Suppose we predict product price based on product rating:

$$ Price = m(Rating) + c $$
Slope

The slope tells us how much the predicted product price changes when the product rating increases by 1 point. For example, if the slope is ₹500, then a 1-point increase in rating is associated with an estimated ₹500 increase in price.

Intercept

The intercept is the predicted price when the product rating is 0. It represents the starting point of the regression line, although a rating of 0 might not be realistic in an actual Flipkart dataset.

# Task 5. Matrix Multiplication in Spotify Recommendation Systems
Prompt

"How does matrix multiplication help in training a machine learning model like Spotify's song recommendation system?"

Summary

Matrix multiplication helps a recommendation system combine information about users and songs. A user-song matrix can represent which songs users have listened to, liked, or interacted with, while another matrix can represent hidden features or preferences. Multiplying these matrices helps the model estimate how strongly a particular user may like different songs. The system can then use these predicted preferences to recommend songs that are likely to match the user's interests.
