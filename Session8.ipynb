```python
# Task 1
def calculate_probability(event_count, total_outcomes):
    return event_count / total_outcomes

# Coin has 1 Head and 2 possible outcomes: Head, Tail
probability_heads = calculate_probability(1, 2)

print("Probability of Heads:", probability_heads)
```

```python
#Task 2
import random

count_six = 0
total_rolls = 100

for i in range(total_rolls):
    roll = random.randint(1, 6)

    if roll == 6:
        count_six += 1

experimental_probability = count_six / total_rolls

print("Number of 6s:", count_six)
print("Experimental Probability:", experimental_probability)
```

```python
# Task 3
genres = [
    'pop', 'rock', 'pop', 'hiphop', 'pop',
    'jazz', 'rock', 'pop', 'hiphop', 'pop',
    'rock', 'pop', 'jazz', 'pop', 'rock',
    'hiphop', 'pop', 'jazz', 'rock', 'pop'
]

pop_count = genres.count('pop')
total_songs = len(genres)

probability_pop = pop_count / total_songs

print("Number of Pop songs:", pop_count)
print("Total songs:", total_songs)
print("Probability of Pop:", probability_pop)
```

```python
# Task 4
def probability_both(food_users, dessert_users):
    food_set = set(food_users)
    dessert_set = set(dessert_users)

    both_users = food_set.intersection(dessert_set)
    all_users = food_set.union(dessert_set)

    if len(all_users) == 0:
        return 0

    probability = len(both_users) / len(all_users)

    return probability


food_users = ['user1', 'user2', 'user3', 'user4', 'user5']
dessert_users = ['user3', 'user4', 'user6', 'user7']

result = probability_both(food_users, dessert_users)

print("Probability of ordering both:", result)
```

```python
# Task 5
phone_users = {
    'user1', 'user2', 'user3', 'user4',
    'user5', 'user6', 'user7', 'user8',
    'user9', 'user10'
}

headphone_users = {
    'user2', 'user3', 'user5', 'user7'
}

both = phone_users.intersection(headphone_users)

probability = len(both) / len(phone_users)

print("Users who bought phone:", len(phone_users))
print("Users who bought both:", len(both))
print("P(Headphones | Phone):", probability)
```
