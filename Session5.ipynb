Simple meaning:

Population = Complete group you want to study.
Sample = Smaller group selected from that population.

```python
# Task 2
import random

# Create 1000 fake user IDs
users = [f"user{i}" for i in range(1, 1001)]

# Select 50 random users
sample_users = random.sample(users, 50)

# Print selected users
print("Selected 50 users:")
for user in sample_users:
    print(user)
```

3. Why Use Sampling for Swiggy Orders?

We use sampling because Swiggy has a huge number of orders, and collecting and analyzing every order would take a lot of time, money, and computational resources. A properly selected sample can give us a good estimate of overall delivery-time patterns much faster.

5. Two Sampling Methods for BookMyShow
A. Random Sampling

Suppose BookMyShow has 1,00,000 users.

Give every user a unique ID.
Randomly select 1,000 users.
Ask those users about their movie booking habits.

Example:
user125, user847, user293...

This gives every user an equal chance of being selected.

B. Stratified Sampling

Divide BookMyShow users into different groups (strata) based on age:

13–17 years
18–30 years
31–50 years
51+ years

Then randomly select users from each group.

Example: If you need 1,000 users, you might select 250 from each age group.
