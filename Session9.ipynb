```python
# Task 1
import numpy as np
import matplotlib.pyplot as plt

# Generate 100 random numbers
data = np.random.normal(loc=50, scale=10, size=100)

print(data)

# Plot histogram
plt.hist(data, bins=10, density=True, edgecolor="black")

plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Normal Distribution (Mean = 50, SD = 10)")
plt.show()
```

```python
# Task 2
import numpy as np
from scipy import stats

steps = [
    5000, 5200, 5100, 5300, 5000,
    5400, 4900, 5100, 5200, 5000,
    5300, 5100, 5000, 5200, 5100,
    5400, 4900, 5000, 5200, 5100
]

mean = np.mean(steps)
median = np.median(steps)
mode = stats.mode(steps, keepdims=False).mode

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

```python
# Task 3
import numpy as np

# Generate scores for 200 players
scores = np.random.normal(
    loc=70,
    scale=15,
    size=200
)

# Count scores between 55 and 85
within_one_sd = np.sum((scores >= 55) & (scores <= 85))

print("Players within one standard deviation:", within_one_sd)
print("Percentage:", (within_one_sd / 200) * 100)
```

# Task 4. Real-World Example — Zomato Delivery Times

Suppose we analyze the delivery times of a large number of Zomato orders in a particular area. If most deliveries take around an average time and fewer orders are extremely faster or slower, the distribution may be approximately bell-shaped. Therefore, a normal distribution can be a useful model for understanding the typical delivery time and estimating how unusual a particular delivery time is. However, real delivery data can be skewed by traffic, weather, distance, and other factors, so normality should be checked rather than assumed.
