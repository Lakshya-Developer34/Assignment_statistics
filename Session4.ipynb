```python
# Task 1
import matplotlib.pyplot as plt
orders = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]

plt.plot(range(1, 11), orders, marker='o')
plt.xlabel("Day")
plt.ylabel("Number of Orders")
plt.title("Daily Swiggy Orders")
plt.show()
```

```python
# Task 2
import statistics

followers = [5, 7, 8, 8, 9, 10, 12, 15, 95]

mean = statistics.mean(followers)
median = statistics.median(followers)

print("Mean:", mean)
print("Median:", median)
```

Task 3 

Suppose a small dataset of video views is:

views = [100, 150, 200, 250, 300, 400, 500, 800, 5000]

Most videos have relatively low views, while one video has 5,000 views.

This creates a long tail toward the right, so the distribution is right-skewed.

Real-life reason: On platforms like YouTube, most videos may receive moderate views, while a few viral videos can receive thousands or millions of views.

```python
# Task 4
import matplotlib.pyplot as plt

ratings = [3, 3, 4, 4, 4, 5, 5, 5, 5, 5]

plt.hist(ratings, bins=[2.5, 3.5, 4.5, 5.5], edgecolor='black')
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.title("Flipkart Product Ratings")
plt.xticks([3, 4, 5])
plt.show()
```
